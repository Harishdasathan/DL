```python
import math

# Given values
real_output = 0.8
fake_output = 0.3

# Calculate Discriminator loss
LD = -(math.log(real_output) + math.log(1 - fake_output))

print("Real image output:", real_output)
print("Fake image output:", fake_output)
print("Discriminator Loss (LD):", round(LD, 4))
```

```python
import math

# Discriminator probability for generated image
D_G_z = 0.2

# Calculate Generator loss
LG = -math.log(D_G_z)

print("Discriminator output for generated image:", D_G_z)
print("Generator Loss (LG):", round(LG, 4))

if LG > 1:
    print("Result: The Generator needs to improve the generated image.")
else:
    print("Result: The Generator is performing reasonably well.")
```

```python
# Image outputs
images = {
    "Real Image": 0.95,
    "Fake Image 1": 0.15,
    "Fake Image 2": 0.75
}

# Classification threshold
threshold = 0.5

print("Threshold:", threshold)
print()

for image, output in images.items():
    if output >= threshold:
        result = "Real"
    else:
        result = "Fake"

    print(image, "-> Output:", output, "-> Classified as:", result)
```

```python
# Given values
total_images = 100
correctly_identified = 80

# Calculate accuracy
accuracy = (correctly_identified / total_images) * 100

print("Total generated images:", total_images)
print("Correctly identified as fake:", correctly_identified)
print("Discriminator Accuracy:", accuracy, "%")
```

```python
# Discriminator output for generated image
D_G_z = 0.9

# Threshold
threshold = 0.5

print("Discriminator output D(G(z)):", D_G_z)

# Classification
if D_G_z >= threshold:
    print("1. Discriminator considers the generated image: REAL")
else:
    print("1. Discriminator considers the generated image: FAKE")

# Check whether result is favorable to Generator
if D_G_z >= threshold:
    print("2. This result is FAVORABLE to the Generator.")
else:
    print("2. This result is NOT favorable to the Generator.")

print("3. Purpose of Generator training:")
print("   To generate realistic images that can fool the Discriminator.")
```
