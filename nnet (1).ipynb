```python
# Simple Neural Network & Logic Gate Simulation

# Import necessary libraries
import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# -----------------------------
# 1. Logic Gate Simulation
# -----------------------------
print("Logic Gate Simulation:")

# Define inputs for 2-input gates
X = np.array([[0,0],[0,1],[1,0],[1,1]])

# AND Gate
AND = np.logical_and(X[:,0], X[:,1]).astype(int)
print("AND Gate Output:", AND)

# OR Gate
OR = np.logical_or(X[:,0], X[:,1]).astype(int)
print("OR Gate Output:", OR)

# NOT Gate (on first input)
NOT = np.logical_not(X[:,0]).astype(int)
print("NOT Gate Output (on first input):", NOT)

# -----------------------------
# 2. Neural Network for AND Gate
# -----------------------------
print("\nNeural Network Training for AND Gate:")

# Target output for AND
y = AND

# Build the neural network
model = Sequential()
model.add(Dense(2, input_dim=2, activation='relu'))  # Hidden layer with 2 neurons
model.add(Dense(1, activation='sigmoid'))           # Output layer

# Compile model
model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])

# Train model
#verbose=0 -> Controls how much information is printed during training:

#0 → Silent, no output, 1 → Progress bar. 2 → One line per epoch.
model.fit(X, y, epochs=1000, verbose=0)

# Predict
predictions = model.predict(X)
predictions = (predictions > 0.5).astype(int)
print("Predicted AND Gate Output by NN:", predictions.flatten())
```
